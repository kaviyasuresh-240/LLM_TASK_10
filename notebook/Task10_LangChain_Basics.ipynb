## Installing LangChain

Required packages:

- langchain
- langchain-core
- langchain-community
- langchain-ollama
- pandas
- jupyter

## Connecting LangChain with Ollama

LangChain was connected to the local Gemma model using Ollama. The OllamaLLM class was used to interact with the model and generate responses locally.

## Creating Prompt Templates

Five healthcare-specific prompt templates were created:

1. General Healthcare
2. Symptoms and Causes
3. Treatment Information
4. Prevention Tips
5. AI in Healthcare

These templates use placeholders to accept dynamic user input.

## Building the Chatbot Chain

A chatbot chain was created using:

- PromptTemplate
- OllamaLLM
- Pipe Operator (|)

The chain accepts a healthcare query and generates an appropriate response using the Gemma model.


## Executing the Chain

The chatbot was tested with healthcare-related questions.

Example:

Question: What is diabetes?

The generated response was displayed in the terminal and saved automatically into a CSV file.

## Summarization Chain

A summarization chain was created to generate concise summaries of healthcare-related text.

The chain accepts input text and produces a summarized output using the Gemma model.

## Question Answering Chain

A Question Answering (QA) chain was developed to answer healthcare-related queries.

Example:

Question: How does AI help doctors?

The chain produces an informative response based on the prompt template.

## Testing with 20 Queries

The chatbot was tested using 20 healthcare-related questions stored in `user_queries.csv`.

The generated responses were saved automatically in `chatbot_responses.csv`.

## Observations

- LangChain simplified prompt management.
- Prompt templates improved response consistency.
- Multiple chains were implemented successfully.
- Responses were generated and stored automatically.

## Strengths

- Runs locally using Ollama and Gemma.
- Supports reusable prompt templates.
- Easy to extend with additional chains.
- No external API dependency.

## Limitations

- Response quality depends on prompt design.
- No external knowledge base is connected.
- Conversation memory is not implemented.
